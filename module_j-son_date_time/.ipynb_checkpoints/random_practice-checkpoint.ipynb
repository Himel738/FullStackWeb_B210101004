# 🔹 Module vs Package in Python

## 🔹 Module
A **module** is a single Python file (`.py`) that contains code such as functions, classes, or variables.

### 🔹 Key Points
- One file  
- Reusable code  
- Easy to import  
- Helps organize small programs  

---

## 🔹 Package
A **package** is a collection of multiple modules organized in a directory (folder).

### 🔹 Key Points
- Folder containing multiple modules  
- Can include sub-packages  
- Used for large projects  
- Helps maintain structure  

---

## 🔹 Difference

| Feature        | Module                | Package                      |
|----------------|----------------------|------------------------------|
| Definition     | Single `.py` file    | Folder of modules            |
| Size           | Small                | Large                        |
| Purpose        | Specific functionality | Organize multiple modules   |
| Structure      | Simple               | Hierarchical                 |

---

## 🔹 Summary

- **Module** → Single file with code  
- **Package** → Collection of modules 

## Random Module

````python
import random
help(random)
````

````python
print(dir(random))
````

````python
print(random.__doc__)
````

````python
print(random.random())  # the value will be given in the range 0.0 <= x < 1.0
print(random.uniform(5,10)) # the value will be given in the range 5.0 <= x < 10.0
print(random.randint(1,10)) # the value will be given in the range 1 <= x <= 10
````

````python
fruits =['Apple', 'Banana', 'Cherry', 'Date', 'Elderberry']
print(random.choice(fruits)) # it will return a random element from the list
random.shuffle(fruits) # it will shuffle the list in place
print(fruits)
````

````python
def generate_pin():
    return random.randint(1000, 9999)
print(f"Your 4 digit pin {generate_pin()}")
````

## 4 - Collection Module: A Special Module where we can find the extended version of list, tuple, Dictionary in built Method

````python
import collections
from collections import defaultdict as dct # We can use as Abbreviation
print(collections.__doc__)
````

````python
print(dir(collections))
````

````python
fruits =['Apple', 'Banana', 'Cherry', 'Date', 'Elderberry', 'Apple']
print(collections.Counter(fruits))
````

## 'Apple' is clearly the most frequent..But for second place, many fruits have same count (1).. Python returns them based on first appearance order

````python
print(collections.Counter(fruits).most_common(2))
````

````python
word_dict = collections.defaultdict(list) # The value will be as a list
word_dict['python'].append('Programming Language')
print(word_dict)
````

## 📌 What is `__name__`?

`__name__` is a special built-in variable in Python. used also for security purpose

It tells whether a file is:
- Executed directly (main program)
- Imported as a module

---

## Date-Time-Module

````python
import datetime as d
now = datetime.datetime.now()
print(now)
````

````python
today_date = d.datetime.now().date()
print(today_date)
````

````python
today_time = d.datetime.now().time()
print(today_time)
````

````python
custom_date_time = d.datetime(2030, 3, 18, 10, 30, 0)
print(custom_date_time)
````

````python
formated_date = now.strftime("%y/%m/%d %H:%M:%S") # If small y 2 digit
print(formated_date)
formated_date = now.strftime("%Y/%m/%d %H:%M:%S")
print(formated_date)
formated_date = now.strftime("%Y/%B/%d %H:%M:%S")
print(formated_date)
formated_date = now.strftime("%Y/%B/%d %A %H:%M:%S")
print(formated_date)
formated_date = now.strftime("%Y/%B/%d %A %I:%M:%S")
print(formated_date)
formated_date = now.strftime("%Y/%B/%d %A %H:%M:%S %p")
print(formated_date)
````

````python
date_str = "25-12-2030 10:45:00"
parsed_date = datetime.datetime.strptime(date_str, "%d-%m-%Y %H:%M:%S")
print(parsed_date)
print(type(parsed_date))
````

## Adding, Subtraction of DateTime

````python
from datetime import datetime,timedelta
today = datetime.today().date()
tommorow = today + timedelta(days = 1)
yesterday = today - timedelta(days = 1)

now = datetime.today()
new_time = now + timedelta(hours = 5, minutes = 30)
print(tommorow)
print(yesterday)
print(new_time)

date1 = datetime(2025, 12, 25)
date2 = datetime(2025, 12, 5)

print(date1 - date2)
````

## TimeZone

````python
import pytz,datetime
from datetime import datetime, UTC
dhaka = pytz.timezone('Asia/Dhaka')
utc = datetime.now(UTC)
print(dhaka)
print(utc)
print(utc.astimezone(dhaka))
print(pytz.all_timezones)
````

````python
import time
start = datetime.now()
time.sleep(5)
end = datetime.now()
print(end - start)
````

# 🔹 JSON File in Python 

## 📌 What is JSON?

JSON stands for **JavaScript Object Notation**.

👉 It is a lightweight data format used for:
- Storing data  
- Exchanging data between systems (API, web apps)

---

## 🔹 Features of JSON

- Easy to read and write  
- Language-independent  
- Uses key-value pairs  
- Similar to Python dictionary  

---

## 🔹 JSON vs Python Dictionary

| JSON | Python |
|------|--------|
| Object | Dictionary |
| Array | List |
| String | str |
| Number | int / float |
| true / false | True / False |
| null | None |

---

## 🔹 JSON Example

```json
{
  "name": "Himel",
  "age": 21,
  "isStudent": true
}


# 🔹 Deserialization in Python (JSON)

## 📌 What is Deserialization?

Deserialization means converting **JSON data → Python object**.

👉 In simple words:  
**Reading JSON and turning it into usable Python data**

---

## 🔹 Why Deserialization?

- To read data from APIs  
- To load data from JSON files  
- To work with external data in Python  

---
